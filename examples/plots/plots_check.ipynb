# Outputs of other libraries

One cell per library whose output the Whybook view shows as the notebook does: a tqdm bar, a Plotly chart, a matplotlib figure, an ipywidgets slider, a plotnine plot, the same plot through ninejs, and an ipympl figure. The kernel's environment needs `pip install -e ".[plots]" matplotlib ipympl plotnine ninejs`. NOTES.md ("Outputs of other libraries") has what the view showed before and after 24 September 2026.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(1)
df = pd.DataFrame({"week": np.repeat(np.arange(20), 10), "arm": np.tile(list("AB"), 100), "pain": rng.normal(5, 1.5, 200)})
df.head(3)
```

```python
import time
from tqdm.auto import tqdm

for _ in tqdm(range(40), desc="bootstrap"):
    time.sleep(0.05)
```

```python
import plotly.express as px

px.scatter(df, x="week", y="pain", color="arm")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(df.week, df.pain, s=8)
ax.set_xlabel("week")
ax.set_ylabel("pain")
```

```python
import ipywidgets as widgets

widgets.IntSlider(value=3, max=10, description="knots")
```

```python
from plotnine import aes, geom_point, ggplot

ggplot(df, aes("week", "pain", color="arm")) + geom_point()
```

```python
from ninejs import interactive

interactive(ggplot(df, aes("week", "pain", color="arm", tooltip="arm")) + geom_point())
```

```python
%matplotlib widget

fig2, ax2 = plt.subplots(figsize=(5, 3))
ax2.plot(df.groupby("week").pain.mean())
```
