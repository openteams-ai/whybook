# Study hours and test scores

Students are nested in schools. Some study hours are missing.

```python
import numpy as np
import pandas as pd


def make_students(n_schools=12, per_school=30, seed=0):
    rng = np.random.default_rng(seed)
    n = n_schools * per_school
    school = np.repeat([f"S{i:02d}" for i in range(n_schools)], per_school)
    school_effect = np.repeat(rng.normal(0, 5, n_schools), per_school)
    hours = rng.gamma(2.0, 2.0, n)
    frame = pd.DataFrame({
        "school": school,
        "sex": rng.choice(["f", "m"], n),
        "hours": hours,
        "score": 50 + 2.5 * hours + school_effect + rng.normal(0, 8, n),
    })
    frame.loc[rng.choice(n, 25, replace=False), "hours"] = np.nan
    return frame


students = make_students()
```
